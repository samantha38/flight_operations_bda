## 3. Missing Value Analysis

```python
import pandas as pd

file = "flights.csv"

# Number of rows processed at a time
chunk_size = 100000

total_rows = 0
missing_values = None

for chunk in pd.read_csv(file, chunksize=chunk_size):

    total_rows += len(chunk)

    # Count missing values in this chunk
    chunk_missing = chunk.isnull().sum()

    if missing_values is None:
        missing_values = chunk_missing
    else:
        missing_values += chunk_missing


print("FULL DATASET INSPECTION")


print("\nTotal rows:", total_rows)

print("\nNumber of columns:", len(missing_values))

print("\nMissing values:")
print(missing_values)
```

## 4. Data Type Analysis

## 5. Duplicate Check

```python
total_rows = 0
duplicate_rows = 0

for chunk in pd.read_csv("flights.csv", chunksize=100000):

    total_rows += len(chunk)
    duplicate_rows += chunk.duplicated().sum()

print("Total rows:", total_rows)
print("Duplicate rows found within chunks:", duplicate_rows)
```

## 6. Invalid Value Check

```python
import pandas as pd

file = "flights.csv"

# Columns needed for validity checks
use_cols = [
    "MONTH",
    "DAY",
    "DAY_OF_WEEK",
    "DISTANCE",
    "DIVERTED",
    "CANCELLED"
]

invalid_counts = {
    "Invalid MONTH": 0,
    "Invalid DAY": 0,
    "Invalid DAY_OF_WEEK": 0,
    "Negative DISTANCE": 0,
    "Invalid DIVERTED": 0,
    "Invalid CANCELLED": 0
}

total_rows = 0

for chunk in pd.read_csv(file, usecols=use_cols, chunksize=100000):
    
    total_rows += len(chunk)

    invalid_counts["Invalid MONTH"] += (
        (~chunk["MONTH"].between(1, 12))
    ).sum()

    invalid_counts["Invalid DAY"] += (
        (~chunk["DAY"].between(1, 31))
    ).sum()

    invalid_counts["Invalid DAY_OF_WEEK"] += (
        (~chunk["DAY_OF_WEEK"].between(1, 7))
    ).sum()

    invalid_counts["Negative DISTANCE"] += (
        chunk["DISTANCE"] < 0
    ).sum()

    invalid_counts["Invalid DIVERTED"] += (
        ~chunk["DIVERTED"].isin([0, 1])
    ).sum()

    invalid_counts["Invalid CANCELLED"] += (
        ~chunk["CANCELLED"].isin([0, 1])
    ).sum()


print("===================================")
print("INVALID VALUE CHECK")
print("===================================")

print("\nTotal rows checked:", total_rows)

for check, count in invalid_counts.items():
    print(f"{check}: {count}")
```

```python
import pandas as pd

file = "flights.csv"

cancelled = 0
diverted = 0
total = 0

for chunk in pd.read_csv(
    file,
    usecols=["CANCELLED", "DIVERTED"],
    chunksize=100000
):
    total += len(chunk)
    cancelled += (chunk["CANCELLED"] == 1).sum()
    diverted += (chunk["DIVERTED"] == 1).sum()

print("===================================")
print("FLIGHT STATUS")
print("===================================")

print("Total flights:", total)
print("Cancelled flights:", cancelled)
print("Diverted flights:", diverted)
print("Completed/non-cancelled flights:", total - cancelled)
```

## keeping categorical cols as strings only

```python
import pandas as pd

input_file = "flights.csv"
output_file = "flights_cleaned.csv"

chunk_size = 100000
first_chunk = True
total_rows = 0

for chunk in pd.read_csv(
    input_file,
    chunksize=chunk_size,
    low_memory=False,
    dtype={
        "ORIGIN_AIRPORT": "string",
        "DESTINATION_AIRPORT": "string",
        "CANCELLATION_REASON": "string"
    }
):
    
    # Write the standardized chunk to the cleaned CSV
    chunk.to_csv(
        output_file,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )
    
    total_rows += len(chunk)
    first_chunk = False


print("PREPROCESSING COMPLETED")

print("Original rows:", 5819079)
print("Rows written:", total_rows)
print("Columns:", len(chunk.columns))
print("Output file:", output_file)
```
